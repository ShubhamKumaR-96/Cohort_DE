```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, count,avg

spark = SparkSession.builder\
       .appName("System_info_Analyser")\
       .master("local[*]")\
       .config("spart.executor.memory","1g")\
       .config("spark.sql.shuffle.partitions","4")\
       .getOrCreate()


print("App_Name: " ,spark.conf.get("spark.app.name"))
print("Shuffle Partitions: ", spark.conf.get("spark.sql.shuffle.partitions"))
print("Master: ", spark.sparkContext.master)

data = [
    ("Alice",   "Engineering", 90000),
    ("Bob",     "Marketing",   55000),
    ("Charlie", "Engineering", 85000),
    ("Diana",   "HR",          48000),
    ("Eve",     "Marketing",   62000),
]

columns = ["name","dept","salary"]

df = spark.createDataFrame(data,columns)


df.show()
```

```python
result = df.groupBy("dept").agg(count("*").alias("headCount"),avg("salary").alias("average_salary"))

result.show()
```

```python
print(f"Parition after suffle: {result.rdd.getNumPartitions()}")
```
